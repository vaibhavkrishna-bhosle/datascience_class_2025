```python
import cv2
```

```python
#read video
cap = cv2.VideoCapture(2)

mirror = True
#mirror the video


while True:
    ret, frame = cap.read()
    if not ret:
        break
    if mirror:
        frame = cv2.flip(frame, 1)
    cv2.imshow('frame', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
cap = cv2.VideoCapture(2)

#crop face using harcascade
face_cascade = cv2.CascadeClassifier('D:\datascience_class_2025\Database\haarcascade_frontalface_default.xml')
while True:
    ret, frame = cap.read()
    if not ret:
        break
    if mirror:
        frame = cv2.flip(frame, 1)
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
    for (x,y,w,h) in faces:
        cv2.rectangle(frame, (x,y), (x+w, y+h), (255,0,0), 2)
    cv2.imshow('frame', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
