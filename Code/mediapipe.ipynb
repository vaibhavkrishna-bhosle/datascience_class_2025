```python
import time
from pathlib import Path
from urllib.request import urlretrieve

import cv2
import mediapipe as mp
from mediapipe.tasks import python as mp_python
from mediapipe.tasks.python import vision

# Download model once (required for Tasks API)
model_path = Path("hand_landmarker.task")
if not model_path.exists():
    model_url = (
        "https://storage.googleapis.com/mediapipe-models/hand_landmarker/"
        "hand_landmarker/float16/1/hand_landmarker.task"
    )
    print("Downloading model...", model_url)
    urlretrieve(model_url, model_path)

# Hand skeleton edges for drawing
HAND_CONNECTIONS = [
    (0, 1), (1, 2), (2, 3), (3, 4),
    (0, 5), (5, 6), (6, 7), (7, 8),
    (5, 9), (9, 10), (10, 11), (11, 12),
    (9, 13), (13, 14), (14, 15), (15, 16),
    (13, 17), (17, 18), (18, 19), (19, 20),
    (0, 17),
]

options = vision.HandLandmarkerOptions(
    base_options=mp_python.BaseOptions(model_asset_path=str(model_path)),
    running_mode=vision.RunningMode.VIDEO,
    num_hands=2,
    min_hand_detection_confidence=0.5,
    min_hand_presence_confidence=0.5,
    min_tracking_confidence=0.5,
 )

cap = cv2.VideoCapture(2, cv2.CAP_DSHOW)
if not cap.isOpened():
    raise RuntimeError("Could not open webcam")

with vision.HandLandmarker.create_from_options(options) as landmarker:
    print("Press 'q' to quit")
    while True:
        ok, frame = cap.read()
        if not ok:
            break

        rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb)
        timestamp_ms = int(time.time() * 1000)

        result = landmarker.detect_for_video(mp_image, timestamp_ms)
        output = frame.copy()

        if result.hand_landmarks:
            h, w = output.shape[:2]
            for hand in result.hand_landmarks:
                pts = []
                for lm in hand:
                    x = int(lm.x * w)
                    y = int(lm.y * h)
                    pts.append((x, y))
                    cv2.circle(output, (x, y), 3, (0, 255, 255), -1)

                for a, b in HAND_CONNECTIONS:
                    cv2.line(output, pts[a], pts[b], (0, 200, 0), 2)

        cv2.imshow("Hand Landmark Detection (MediaPipe Tasks)", output)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

cap.release()
cv2.destroyAllWindows()
```
