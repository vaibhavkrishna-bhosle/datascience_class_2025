```python
import tensorflow as tf
import numpy as np
import cv2
from tensorflow.keras.preprocessing import image
import pickle

IMG_SIZE = (224, 224)

# Load model
with open(r"D:\satyam apps\datascience_class_2025\classification_model\resnet50_model.pkl", "rb") as f:
    resnet_model = pickle.load(f)

CLASS_NAMES = ["Cat", "Dog"]

def preprocess_frame(frame):
    frame_resized = cv2.resize(frame, IMG_SIZE)
    img_array = image.img_to_array(frame_resized)
    img_array = np.expand_dims(img_array, axis=0)
    return img_array

def predict_frame(model, frame):
    img = preprocess_frame(frame)
    pred = model.predict(img, verbose=0)[0][0]
    label = CLASS_NAMES[int(pred > 0.5)]
    return label, float(pred)

# Start webcam
cap = cv2.VideoCapture(2)

if not cap.isOpened():
    raise RuntimeError("Could not open webcam.")

print("Press 'q' to quit webcam window.")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    label, conf = predict_frame(resnet_model, frame)

    text = f"Prediction: {label} | Confidence: {conf:.4f}"
    cv2.putText(frame, text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

    cv2.imshow("Live Pet Classifier", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
